# Coding Question 6: Factorial of a Number

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(4))  
```

# Coding Question 12: Check Palindrome (String)

```python
s = input("Enter a string:")

if s == s[::-1] :
    print("palindrome string")

else :
    print ("Not a palindrome")
```

# Palindrome Number 

```python
num = int (input("Enter A number"))
original = num
rev = 0

while num > 0 :
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if(original == rev):
    print ("Number is palindrome")

else :
    print("Number is not palindrome") 
    
```

# Coding Question 13: Find Maximum and Minimum in a List

```python
num = [10,45,3,99,27]

largest = num[0]
smallest = num[0]

for n  in num :
    if n > largest :
        largest = n 

    if n < smallest:
        smallest = n

print("largest no is",largest)
print("smallest no is",smallest)


    
```
