```python
# zip() is a built-in Python function that combines two or more iterables into a single iterator of tuples by pairing elements at the same index.

#Because zip() does not return a list. It returns a zip object (an iterator).


subjects = ['Math', 'Python', 'Java', 'C++']
marks = [50, 60, 40, 80]

z = zip(subjects, marks)
print(z)
```

```python
# zip return zip object iterator not support indexing ....

subjects = ['Math', 'Python', 'Java', 'C++']
marks = [50, 60, 40, 80]

z = zip(subjects, marks) 
print(z[0])
```

```python
# to support indexing we need to convert it to the list
subjects = ['Math', 'Python', 'Java', 'C++']
marks = [50, 60, 40, 80]

z = list (zip(subjects, marks) )
print(z)
```
