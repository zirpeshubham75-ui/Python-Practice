## count a how many iteration happen in code for sepefic element in list 

```python
l = [11,23,45,88,66,67,98,78,25] 
num = int (input("enter the number from list :"))
count = 0
for i  in range(len(l)):
   count+=1
   if(num == l[i]):
       print("Number is present")
      
print("No of iteration :",count)        
```

## to stop loop use BREAK KEYWORD

```python
l = [11,23,45,88,66,67,98,78,25] 
num = int (input("enter the number from list :"))
count = 0
for i  in range(len(l)):
   count+=1
   if(num == l[i]):
       print("Number is present")
       break
print("No of iteration :",count) 
```

## given a list of numbers ,find the frequency of occurence of distinct numbers ..

```python
l = [11,11,33,33,44,33,11,33,55,55]
freq = {} ;
for i in l  :
    if i in freq.keys() :
        freq[i]+=1
    else :
        freq[i]=1 ;
print(freq)

for key,value in freq.items():
 print( key,"occure ",value,"times")
    
    
```

```python
l = [11,11,33,33,44,33,11,33,55,55]
largest=l[0]
for i in l :
    if i > largest :
       largest = i

print("the largest number from the list is :",largest)    
```
