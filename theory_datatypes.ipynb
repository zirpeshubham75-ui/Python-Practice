# append() vs extend()
 - Both methods are used to add elements to a list. append() adds the entire object as a single element, whereas extend() adds each element of an iterable individually.

# remove() vs pop() vs del
 -All three are used to remove elements, but they work differently. remove() deletes by value, pop() deletes by index and returns the removed element, while del is a Python statement that deletes elements, slices, or even entire variables.

# sort() vs sorted()
 -Both are used for sorting. sort() sorts the original list in place and returns None, whereas sorted() returns a new sorted list without modifying the original iterable.

# reverse() vs [::-1]
 -reverse() reverses the original list in place, whereas slicing with [::-1] creates and returns a new reversed copy without modifying the original list.

 # Shallow Copy vs Deep Copy
 -A shallow copy creates a new outer object, but nested mutable objects are still shared between the original and the copy. A deep copy creates completely independent copies of both the outer object and all nested objects.

#  String 

 ## 1 What is a String?


 -A string is an immutable sequence of Unicode characters used to store  in Python. Strings are enclosed in single quotes (' '),      double quotes (" "), or triple quotes (''' ''' or """ """)

 ## 2 Is a String Mutable?

-No. Strings are immutable in Python. Once a string is created, its characters cannot be modified. Any operation that appears to modify a string actually creates a new string object.

 ## 3 Why are Strings Immutable?


-Strings are immutable to improve memory efficiency, performance, and security. Since they cannot change, Python can safely reuse the same string object in memory and use strings as dictionary keys and set elements.

## 6 Difference Between find() and index()

## 7 Difference Between split() and join()
Interview Answer

split() converts a string into a list by breaking it at a specified separator, whereas join() combines elements of an iterable into a single string using a specified separator.

## 8 split() vs partition()
Interview Answer

Both split() and partition() are used to divide a string. The difference is that split() breaks a string into multiple parts and returns a list, whereas partition() splits the string only at the first occurrence of the separator and returns a tuple containing three elements.

# Dictionay 
## 1 What is a Dictionary?
Interview Answer

A dictionary is a mutable mapping data type in Python that stores data as key-value pairs. Each key must be unique and hashable, while values can be of any data type and may be duplicated.

## 2 Why are Dictionary Keys Unique?
Interview Answer

Dictionary keys are unique because each key identifies exactly one value. If duplicate keys were allowed, Python would not know which value to return.

## 3 Difference Between get() and []
Interview Answer

Both get() and square brackets ([]) are used to access dictionary values. The difference is that get() returns None (or a default value) if the key is missing, whereas [] raises a KeyError.

## 4 Why can't a List be a Dictionary Key?
Interview Answer

A list cannot be used as a dictionary key because lists are mutable and therefore not hashable. Dictionary keys must be hashable so that Python can compute a stable hash value for fast lookups. If a key could change after insertion, the dictionary would no longer be able to locate it correctly

# What is None?
Interview Answer

None is a special constant in Python that represents the absence of a value or a null value. It is the only instance of the NoneType class and is commonly used as a default return value or placeholder

## What is Dynamic Typing?
Interview Answer

Python is a dynamically typed language, which means you do not need to declare the data type of a variable explicitly. The data type is determined automatically at runtime based on the value assigned to the variable.
