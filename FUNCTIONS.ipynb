# DEFAULT PARAMETERS
 -A default parameter is a parameter that already has a value assigned in the function definition. If no argument is passed while calling the function, Python uses the default value

## Why do we use default parameters?



-We use default parameters to make function arguments optional. If the user doesn't provide a value, the function uses the default value.

## Where should default parameters be placed?
  -Default parameters must come after non-default parameters because Python assigns arguments from left to right

```python
def divide(num1,num2=2):
    result = num1/num2
    return result

divide(56,4)    
  
```

```python
divide(56)    # it will take default argument
```

## POSITIONAL ARGUMENTS AND KEYWORDS ARGUMENTS
  -Positional arguments are matched to function parameters based on their position (order)

  -Keyword arguments are matched to parameters by their parameter names, not by their position.

```python
def deatils(name,age,city):        ##POSITIONAL ARGUMENTS
   print("NAME : ",name)
   print("AGE : ", age)

   print("CITY : ",city)
deatils("sahil",23,'pune')


```

```python
deatils(23,'pune','sahil')
```

```python
deatils(age=23,city='pune',name='sahil')      ## KEYWORDS ARGUMENTS
```

##  ARGUMENTS AND KEYWORDS ARGUMENTS  (ARGS AND KWAEGS ) 

  - *args allows a function to accept any number of positional arguments.

  - **kwargs allows a function to accept any number of keyword arguments.

```python
def add_no(num1 , num2):
    result = num1 +num2
    return result
    
add_no(2,3,4)
```

```python
# taking mmultiple argument using *
def add_no(*nums) :
    result =0
    for i in nums :
        result+=i
    return result   
```

```python
add_no(3)
```

```python
add_no(3,4,4,4)
```

```python
add_no(3,2)
```
