## IF ELIF ELSE 
  - ELIF IS USED WHEN MULTIPLE CONDITIONES ARE THERE ..
  - ONLY ONE IF AND ELSE SHOULD BE THERE. 
  - THERE CANN BE MULTIPLE ELIF.

```python
a = int ( input('first no :'))
b=  int (input('second no :'))
op = input("Enter operation (+/-) :")

if op == '+':
    print('the addition of ',a ,'and',b,'is :',a+b)
elif (op == '-'):
        print('the subtration of ',a ,'and',b,'is :',a-b)

else :
    print ("this operation is not allowed :")
     
     
```

```python
a = int ( input('first no :'))
b=  int (input('second no :'))
op = input("Enter operation ( ) :")

if op == '+':
    print('the addition of ',a ,'and',b,'is :',a+b)
elif (op == '-'):
        print('the subtration of ',a ,'and',b,'is :',a-b)
elif (op == '*'):
        print('the multi of ',a ,'and',b,'is :',a*b)

elif (op == '/'):
        print('the division of ',a ,'and',b,'is :',a/b)

elif (op == '//'):
        print('the floor division of ',a ,'and',b,'is :',a//b)

elif (op == '**'):
        print('the ** of ',a ,'and',b,'is :',a**b)
   

else :
    print ("this operation is not allowed :")
```
